```python
import pandas as pd
from matplotlib import pyplot as plt 
import seaborn as sns
```

```python
titanic = pd.read_csv('titanic.csv')
```

```python
titanic.head()
```

```python
sns.barplot(x = 'Pclass', y = 'Survived', data = titanic, hue = 'Pclass', legend = False, palette = 'Set2')
plt.title('Percentage of survivors by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel("Survival percentage")
plt.savefig('Percent_Pclass.jpg')
```

```python
sns.countplot(x = 'Pclass', hue = 'Survived', data = titanic)
plt.title('Number of survivors and non-survivors by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel("Number of people")
plt.savefig('Number_Pclass.jpg')
```

```python
sns.barplot(x = 'Sex', y = 'Survived', data = titanic, hue = 'Sex', legend = False, palette = ['#84B6F4',"#FF6961"])
plt.title('Percentage of survivors by Passenger Sex')
plt.xlabel('Passenger Sex')
plt.ylabel("Survival percentage")
plt.savefig('Percent_Sex.jpg')
```

```python
sns.countplot(x = 'Sex', hue = 'Survived', data = titanic)
plt.title('Number of survivors and non-survivors by Passenger Sex')
plt.xlabel('Passenger Sex')
plt.ylabel("Number of people")
plt.savefig('Number_Sex.jpg')
```

```python
sns.barplot(x = 'Age', y = 'Survived', data = titanic)
plt.savefig('Fail_Age.jpg')
```

```python
# Create age range bins
titanic['Age_Range'] = pd.cut(titanic['Age'], bins=[0, 18, 30, 50, 100], 
                              labels=['Children(0-18)', 'YoungAdults(19-30)', 'Adults(31-50)', 'Seniors(51+)'])

# Create the barplot with age ranges
sns.barplot(x='Age_Range', y='Survived', data=titanic)
plt.title('Percentage of survivors by Passenger Age')
plt.xlabel('Passenger Age Range')
plt.ylabel("Survival Percentage")
plt.savefig('Percent_Age.jpg')
```

```python
sns.countplot(x='Age_Range', hue='Survived', data=titanic)
plt.title('Number of survivors and non-survivors by Passenger Age')
plt.xlabel('Passenger Age Range')
plt.ylabel("Number of people")
plt.savefig('Number_Age.jpg')
```
